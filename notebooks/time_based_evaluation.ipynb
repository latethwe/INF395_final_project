# Time-Based Evaluation Pipeline

Этот ноутбук запускает новый time-based pipeline:
1. `scripts/06_time_split.py` — сплит по `collected_at`
2. `scripts/07_evaluate_time_based.py` — обучение и метрики MAE/RMSE


```python
from pathlib import Path
import json
import subprocess

PROJECT_ROOT = Path('/Users/zhasik/Desktop/krisha')
PARQUET = PROJECT_ROOT / 'data/index/index.parquet'

assert PARQUET.exists(), f'Parquet not found: {PARQUET}'
print('Dataset:', PARQUET)
```

```python
cmd = [
    'python', 'scripts/06_time_split.py',
    '--parquet', 'data/index/index.parquet',
    '--out-dir', 'data/processed',
    '--train-ratio', '0.8',
    '--val-ratio', '0.1',
]
print('RUN:', ' '.join(cmd))
subprocess.run(cmd, cwd=PROJECT_ROOT, check=True)
```

```python
cmd = [
    'python', 'scripts/07_evaluate_time_based.py',
    '--parquet', 'data/index/index.parquet',
    '--splits-dir', 'data/processed',
    '--out-json', 'data/processed/time_eval_metrics.json',
]
print('RUN:', ' '.join(cmd))
subprocess.run(cmd, cwd=PROJECT_ROOT, check=True)
```

```python
metrics_path = PROJECT_ROOT / 'data/processed/time_eval_metrics.json'
metrics = json.loads(metrics_path.read_text(encoding='utf-8'))
metrics
```
